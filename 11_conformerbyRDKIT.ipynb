```python
from rdkit import Chem
from rdkit.Chem import AllChem

# 1. Define caffeine molecule
caffeine = Chem.MolFromSmiles("CN1C=NC2=C1C(=O)N(C(=O)N2C)C")
caffeine = Chem.AddHs(caffeine)

# 2. Generate 10 conformers using RDKit's ETKDG algorithm
conf_ids = AllChem.EmbedMultipleConfs(caffeine, numConfs=10, randomSeed=42)

# 3. Optimize each conformer and get its energy (MMFF94 force field)
energies = []
for cid in conf_ids:
    AllChem.MMFFOptimizeMolecule(caffeine, confId=cid)
    ff = AllChem.MMFFGetMoleculeForceField(caffeine, AllChem.MMFFGetMoleculeProperties(caffeine), confId=cid)
    energies.append(ff.CalcEnergy())

# 4. Print results
for i, e in enumerate(energies):
    print(f"Conformer {i}: Energy = {e:.4f} kcal/mol")

# 5. Save all conformers to a single SDF file
writer = Chem.SDWriter("caffeine_etkdg_conformers.sdf")
for cid in conf_ids:
    writer.write(caffeine, confId=cid)
writer.close()
print("\nSaved 10 conformers to caffeine_etkdg_conformers.sdf")
```

### CODE TO VISUALIZE CONFORMERS ONE AT A TIME

```python
from ipywidgets import interact
!pip install py3Dmol -q
import py3Dmol
def show_conf(i=0):
    v = py3Dmol.view(width=450, height=400)
    v.addModel(Chem.MolToMolBlock(caffeine, confId=conf_ids[i]), "mol")
    v.setStyle({'stick': {}, 'sphere': {'scale': 0.25}})
    v.zoomTo()
    v.show()

interact(show_conf, i=(0, len(conf_ids) - 1))
```
